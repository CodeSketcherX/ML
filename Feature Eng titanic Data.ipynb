```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
import pandas as pd

df = pd.read_csv(r"path")

df.head()
```

```python
print("First five records:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nDataset information:")
df.info()

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Check Missing Value
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

```python
age_median = df["Age"].median()

df["Age"] = df["Age"].fillna(age_median)

print("Median Age:", age_median)
print("Missing values in Age:", df["Age"].isnull().sum())
```

```python
embarked_mode = df["Embarked"].mode()[0]

df["Embarked"] = df["Embarked"].fillna(embarked_mode)

print("Mode of Embarked:", embarked_mode)
print("Missing values in Embarked:", df["Embarked"].isnull().sum())
```

```python
label_encoder = LabelEncoder()

df["Sex"] = label_encoder.fit_transform(df["Sex"])

print("Sex column after Label Encoding:")
display(df[["Sex"]].head(10))

print("\nEncoding:")
print(dict(zip(label_encoder.classes_,
               label_encoder.transform(label_encoder.classes_))))
```

```python
df = pd.get_dummies(
    df,
    columns=["Embarked"],
    prefix="Embarked",
    dtype=int
)

print("Dataset after One-Hot Encoding:")
display(df.head())

print("\nOne-Hot Encoded columns:")
print([col for col in df.columns if col.startswith("Embarked_")])
```

```python
numerical_columns = ["Age", "SibSp", "Parch", "Fare"]

scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(df[numerical_columns])

print("Dataset after Feature Scaling:")
display(df[numerical_columns].head())
```

```python
print("First five records after feature engineering:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Missing values after feature engineering:")
print(df.isnull().sum())

print("\nTotal remaining missing values:")
print(df.isnull().sum().sum())
```

```python
print("Sex values:")
print(df["Sex"].unique())

print("\nEmbarked encoded columns:")
display(df.filter(regex="^Embarked_").head())

print("\nData types:")
print(df.dtypes)
```

```python
df.to_csv("titanic_cleaned.csv", index=False)

print("Cleaned dataset saved successfully as titanic_cleaned.csv")
```
